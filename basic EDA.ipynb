#### Single Variable plots

### Single-Variable Plots

Explore numerical features to understand their distributions, ranges, and potential outliers.

- **Histograms:** Visualize the distribution of numerical variables and identify skewness.
- **Box plots:** Examine the spread of values and detect potential outliers.
- **Categorical plots:** Explore the frequency of different categories.

These visualizations help identify patterns and data quality issues. If a variable is highly skewed, transformations may be useful for certain statistical analyses or machine learning models, depending on the algorithm and the task.

```python
df_agg.Likes.hist(bins=100)
```

The distribution of Likes is strongly right-skewed. Most observations have relatively low like counts, while a small number have significantly higher values. One observation exceeds 200,000 likes, which may be an outlier worth investigating further.

```python
plt.hist(df_agg['Av­er­age per­cent­age viewed (%)'])
```

Most observations are concentrated in the middle of the distribution, with the highest frequency around 35–40. The frequency gradually decreases toward higher values, with only a few observations at the upper end of the range.

```python
df_agg['Im­pres­sions click-through rate (%)'].hist()
```

The distribution is right-skewed, with most observations concentrated between 1 and 5 and a few unusually high values. We may consider applying a transformation later if needed for linear regression or other statistical methods.

```python
plt.boxplot(df_agg['Likes'])
```

The box plot reveals several potential outliers in the Likes column, including values much higher than the majority of observations. These extreme values may influence the mean, so the median could provide a more representative measure of a typical post's likes. The outliers should be investigated before deciding whether any data cleaning or transformation is necessary.

```python
plt.boxplot(df_agg['Av­er­age per­cent­age viewed (%)'])
```

The box plot shows that most observations fall between approximately 25-45. The distribution appears relatively balanced compared with the Likes variable. However, a few potential outliers remain above the upper whisker.

```python
plt.boxplot(df_agg['Im­pres­sions click-through rate (%)'])
```

The box plot shows that most observations fall between approximately 2 and 4, with a median around 3. Several potential outliers appear above the upper whisker, including one particularly high value. This suggests that most observations have relatively low values, while a few are substantially higher.

```python
bins = pd.IntervalIndex.from_tuples([
    (0, 100),
    (100, 1000),
    (1000, float("inf"))
])

df_agg['rev_buckets'] = pd.cut(
    df_agg['Your es­tim­ated rev­en­ue (USD)'],
    bins
)
```

The bar chart shows the distribution of videos across three revenue categories. Most videos generate less than \$100, around 50 fall between \$100 and \$1,000, and only a few exceed \$1,000. This indicates that the dataset is heavily concentrated in the lowest revenue category.

#### Relationshipd and Multi_variable Plots

### Relationships and Multivariable Plots

Multivariable EDA explores relationships between two or more variables to identify patterns, associations, and potential predictive features.

We will use several visualization techniques:

- **Scatter plots:** Examine relationships between two numerical variables.
- **Correlation matrices:** Summarize the strength and direction of correlations between numerical variables.
- **Pivot tables:** Aggregate data to compare groups.
- **Bar charts:** Compare values or frequencies across categories.
- **Line charts:** Identify trends and changes, particularly over time.

These techniques help us understand how variables relate to each other and identify patterns that may be useful for machine learning.

Scatter Plots \
**Positive relationship**: Both variables tend to increase together. \
**Negative relationship**: One tends to decrease as the other increases. \
**No clear relationship**: The points show no obvious pattern.\
**Outliers**: Some videos behave very differently from the rest.

```python
sns.regplot(x="Av­er­age per­cent­age viewed (%)", y="CPM (USD)", data=df_agg)
```

The scatter plot suggests a weak positive relationship between average percentage viewed and CPM. The trend line slopes slightly upward, but the points are widely dispersed, indicating that the relationship is not strong. A few observations have unusually high CPM values. This suggests that average percentage viewed alone may not be a strong predictor of CPM.

```python
#RPM - how much the youtuber makes for an add. CPM - how much youtuber sells the ad for
sns.regplot(x='RPM (USD)', y='CPM (USD)', data = df_agg)
```

The scatter plot shows a positive relationship between RPM and CPM. The upward trend is more pronounced than in the previous plot, suggesting a stronger association between the two variables. A few observations have unusually high CPM values and may influence the regression line.

Correlation Matrices \
Scatter plots are great for comparing two variables. Often we have many different variables that we want to see relationships between.

In this case we would use a correlation matrix.

```python

# Calculate correlations only between numeric columns
corr = df_agg.corr(numeric_only=True)

# Plot the correlation heatmap
sns.heatmap(corr)
```

```python

sns.set_theme(style="white")

corr = df_agg.corr(numeric_only=True)

# Hide the upper triangle and diagonal
mask = np.triu(np.ones_like(corr, dtype=bool))

# Create the figure
f, ax = plt.subplots(figsize=(12, 10))

# Color palette
cmap = sns.diverging_palette(230, 20, as_cmap=True)

# Draw the heatmap
sns.heatmap(
    corr,
    mask=mask,
    cmap=cmap,
    center=0,
    square=True,
    linewidths=0.5,
    cbar_kws={"shrink": 0.8},
    annot=True,
    fmt=".2f",
    annot_kws={"fontsize": 7},
    ax=ax
)

plt.tight_layout()
plt.show()
```

**Correlation Heatmap**

A correlation heatmap helps us examine the linear relationships between multiple numerical variables simultaneously.

**Key observations:**
- Views, watch time, engagement metrics, impressions, and estimated revenue show very strong positive correlations.
- RPM and CPM have a moderate positive correlation of approximately 0.50.
- Average percentage viewed and impressions click-through rate show weak linear correlations with most other variables.

**Purpose:** This visualization helps identify relationships, potential feature redundancy, and patterns worth investigating further. Correlation does not imply causation, and the observed relationships may depend on the dataset's structure.

### Pivot Tables
Sometimes we want to cut the data and compare how variables stack up. For example we may want to see which countries have the highest view duration. We can do this easily with pivot tables.

We will use the df_agg_country_sub dataset to explore this.

```python
#  A basic pivot table takes the average of these different categories we choose.
pd.pivot_table(
    df_agg_country_sub,
    index='Country Code',
    values='Average View Percentage'
).sort_values('Average View Percentage')
```

```python
#this time, let's compare average view percentage by country code and subscriber status.
pd.pivot_table(
    df_agg_country_sub,
    index='Country Code',
    columns='Is Subscribed',
    values='Average View Percentage'
)
#we could plot this to see which countries have the biggest difference 
# in subscribed watch time for subscribed vs not subscribed viewers 
```

Purpose: To compare viewing behavior between subscribed and non-subscribed viewers across countries and identify differences that may deserve further investigation.

Missing values (NaN) indicate that an average is unavailable for a particular group. These results show associations, not necessarily causal relationships.

```python
pd.pivot_table(df_agg_country_sub, index = 'Is Subscribed', values = 'Average View Percentage').plot.bar()
```

The bar chart shows that subscribers watch a higher average percentage of videos than non-subscribers (approximately 37% vs. 33%). This suggests that subscribers tend to watch a larger proportion of each video. However, this comparison alone does not establish causation or tell us the absolute watch time.

### Line Charts

Line charts visualize how a numerical variable changes over time.

We will use the df_ts dataset for this analys.



```python
# Fisrt, converting the dstring to datetime format
# make sure tha dat field is in the date time format
df_ts['Date'] = pd.to_datetime(
    df_ts['Date'],
    format='mixed',
    dayfirst=True
)
#let's look at user subscriptions removed and see if there is any trend there. 
# We will also compare this with user likes removed to see if we can find anything interesting

# aggregate these by video by using pivot table
rm_x_date = pd.pivot_table(df_ts, index = 'Date', values ='User Subscriptions Removed', aggfunc = 'sum').reset_index()

# visualize the data with seaborn
sns.lineplot(data = rm_x_date, x = 'Date', y = 'User Subscriptions Removed')

```

```python
#now let's compare this with two other related metrics, video likes removed and dislikes

likes_rm_date = pd.pivot_table(df_ts, index = 'Date', values ='Video Likes Removed', aggfunc = 'sum').reset_index()
dislikes_rm_date = pd.pivot_table(df_ts, index = 'Date', values ='Video Dislikes Added', aggfunc = 'sum').reset_index()

sns.lineplot(data = rm_x_date, x = 'Date', y = 'User Subscriptions Removed')
sns.lineplot(data = likes_rm_date, x = 'Date', y = 'Video Likes Removed')
sns.lineplot(data = dislikes_rm_date, x = 'Date', y = 'Video Dislikes Added')   
```

```python
# thats a bit mess, we are gonna compare it over months instead of days 
df_ts['YearMonth'] = df_ts['Date'].dt.to_period('M')

# create new pivot tables
rm_x_date = pd.pivot_table(df_ts, index='YearMonth', values='User Subscriptions Removed', aggfunc='sum').reset_index()
likes_rm_date = pd.pivot_table(df_ts, index='YearMonth', values='Video Likes Removed', aggfunc='sum').reset_index()
dislikes_rm_date = pd.pivot_table(df_ts, index='YearMonth', values='Video Dislikes Added', aggfunc='sum').reset_index()

# create 3 separate line plots with pandas built in vusialization 
rm_x_date.plot(x='YearMonth', y='User Subscriptions Removed', title='User Subscriptions Removed Over Time')
likes_rm_date.plot(x='YearMonth', y='Video Likes Removed', title='Video Likes Removed Over Time')
dislikes_rm_date.plot(x='YearMonth', y='Video Dislikes Added', title='Video Dislikes Added Over Time')
```

```python
# This was a bit messy, let's compare it over months instead of days. 
# To get everything on one graph, we need to make it so every video has the same day.
# Just different months and years 

df_ts['YearMonth'] = df_ts['Date'].apply(lambda x: x.replace(day=1))

rm_x_date = pd.pivot_table(df_ts, index='YearMonth', values='User Subscriptions Removed', aggfunc='sum').reset_index()
likes_rm_date = pd.pivot_table(df_ts, index='YearMonth', values='Video Likes Removed', aggfunc='sum').reset_index()
dislikes_rm_date = pd.pivot_table(df_ts, index='YearMonth', values='Video Dislikes Added', aggfunc='sum').reset_index()

# creat 3 line plots with seaborn all on the same graph
sns.lineplot(data=rm_x_date, x='YearMonth', y='User Subscriptions Removed', label='User Subscriptions Removed')
sns.lineplot(data=likes_rm_date, x='YearMonth', y='Video Likes Removed', label='Video Likes Removed')
sns.lineplot(data=dislikes_rm_date, x='YearMonth', y='Video Dislikes Added', label='Video Dislikes Added')
```

```python
#Let's normalize this by views 

df_ts['Month_Year'] = df_ts['Date'].apply(lambda x: x.replace(day=1))

monthly_views = pd.pivot_table(df_ts, index = 'Month_Year', values = 'Views', aggfunc = 'sum')
rm_x_date = pd.pivot_table(df_ts, index='Month_Year',values = 'User Subscriptions Removed', aggfunc ='sum') / monthly_views.values
likes_rm_date = pd.pivot_table(df_ts, index='Month_Year',values='Video Likes Removed', aggfunc = 'sum') / monthly_views.values
dislikes_date = pd.pivot_table(df_ts, index='Month_Year',values='Video Dislikes Added', aggfunc = 'sum') / monthly_views.values

#create 3 line plots with seaborn all on one graph 
sns.lineplot(data=rm_x_date,x='Month_Year', y='User Subscriptions Removed', label ='Subs Removed')
sns.lineplot(data=likes_rm_date,x='Month_Year', y='Video Likes Removed', label = 'Likes Removed')
sns.lineplot(data=dislikes_date,x='Month_Year', y='Video Dislikes Added', label = 'Dislikes')

#we can see that there was still a spike in the normalized metrics during the same point in the year. It is worth exploring this further! 
```

Which videos get the most views?

```python
# Remove the summary row and keep individual videos
videos = df_agg[df_agg['Video title'].notna()].copy()
videos = videos[videos['Video'] != 'Total']

top_10 = videos.nlargest(10, 'Views').sort_values('Views')

plt.figure(figsize=(12, 7))
sns.barplot(data=top_10, x='Views', y='Video title')
plt.title('Top 10 Most-Viewed Videos')
plt.xlabel('Views')
plt.ylabel('Video title')
plt.tight_layout()
plt.show()
```

Hypothesis: Practical, beginner-oriented content and career guidance may be appealing to this channel's audience. However, views alone cannot tell us whether the topic, title, thumbnail, video age, or external promotion explains the performance.

Do more views mean better audience retention?

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=videos,
    x='Views',
    y='Average percentage viewed (%)'
)

plt.title('Views vs Average Percentage Viewed')
plt.xlabel('Views')
plt.ylabel('Average Percentage Viewed (%)')
plt.tight_layout()
plt.show()
```

Insight: There is no obvious strong relationship between views and average percentage viewed. \
Most videos have relatively low view counts but span a wide range of retention values, from roughly 5% to 75%. The highest-viewed video has an average percentage viewed of approximately 37%. \
This suggests that getting more views does not necessarily mean retaining a larger proportion of viewers

```python
plt.figure(figsize=(9, 6))

sns.regplot(
    data=videos,
    x='Impressions click-through rate (%)',
    y='Average percentage viewed (%)',
    scatter_kws={'alpha': 0.5}
)

plt.title('CTR vs Average Percentage Viewed')
plt.xlabel('Impressions Click-Through Rate (%)')
plt.ylabel('Average Percentage Viewed (%)')
plt.tight_layout()
plt.show()
```

The scatterplot shows a weak positive trend, with considerable variation between videos. A higher CTR does not consistently correspond to higher retention.
